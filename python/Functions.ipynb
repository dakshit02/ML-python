```python
def odd_even(NUM):
    if NUM % 2 == 0:
        print(f"{NUM} is even")
    else:
        print("odd")
        

#calling the function
odd_even(NUM)
```

```python
# multiple parameter
def add(a, b):
    c = a + b
    return c


new_var = add(2, 4)

print(new_var)
```

Arguments 

```python
# Variable Length Arguments

def print_number(*args):
    for numbers in args:
        print(numbers)


print_number(1, 2, 3, 4, 5, 6, "dakshit")
```

```python
# Keyword Argument


def print_details(name="dakshit", age=23):
    print(f"My name is {name}, and my age is {age}")



print_details()
```

```python
# multiple keyword arguments
def print_details_new(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")


print_details_new(name="dakshit", age=23, gender="M")
```

```python
# using * args & ** kwargs together


def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional Argument = {val}")

    print("Key Word Argument")

    for key, value in kwargs.items():
        print(f"{key}:{value}")


print_details(1, 2, 3, 4, "dakshit", name="Ayush", age=23, gender="M")
```
