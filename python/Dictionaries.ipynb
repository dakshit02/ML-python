```python
#creating dictionary 
student = {"name":"Dakshit",
           "age":23,
           "surname":"chandel"}


print(student)
```

```python
#acessing dictionary items 
print(student["name"])
print(student["age"])


#accessing with get method 
print(student.get('surname'))
print(student.get("name"))
```

```python
#single key is always used 
student_1={"name":"Dakshit",
           "age":23,
           "name":"chandel"}
print(student_1)
```

```python
##modifying dictonary elements
##dictonary are mutable , so you can add , update or delete element 
print(student)
```

```python
# updating a key value
student["age"] = 33
print(student)
```

```python
# adding new key value
student["address"] = "India"
print(student)
```

```python
# deleting a key value pair
del student["age"]
print(student)
```

```python
# Dictonary Methods
keys = student.keys()  ## gets all the keys
print(keys)
values = student.values()  ##gets all the values
print(values)

items = student.items()  ##gets all key value pairs
print(items)
```

Shallow Copy


```python
# shallow copy - we use this to copy a distonary as a disticnt dictionary
dict_1 = {"name": "daku", "age": "23"}
dict_2 = dict_1

print(dict_1)
print(dict_2)

# changin the value of a key oin 1st dict chnage its value in the second dict as well

dict_1["name"] = "daku_1"

print(dict_1)
print(dict_2)
```

```python
# so to avoid the above thing to happen we use shallow copy
dict_3 = dict_1.copy()  # shallow copy
dict_1["name"] = "daku_2"

print(dict_1)
print(dict_3)
```

Iterating over Dictionaries

```python
# iterating on keys
for i in student.keys():
    print(i)
```

```python
# iterating on values
for j in student.values():
    print(j)
```

```python
# iterating on both key and values
for i, j in student.items():
    print(f"{i} : {j}")
```

```python
# nested Dictionary
nest_dict = {
    "stud_1": {"name": "daku", "age": 23},
    "stud_2": {"name": "ayush", "age": 24},
}

print(nest_dict)
```

```python
print(nest_dict["stud_1"])
```

```python
# acess a specific value
print(nest_dict["stud_2"]["name"])
```

```python
print(nest_dict.items())
```

```python
print(nest_dict["stud_1"].items())
```

```python
# looping through the nested dictionary
for i,j in nest_dict.items():
    print(f"{i}:{j}")
    for key, value in j.items():
        print(f"{key}:{value}")
```
