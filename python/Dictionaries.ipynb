```python
#creating dictionary 
student = {"name":"Dakshit",
           "age":23,
           "surname":"chandel"}


print(student)
```

```python
#acessing dictionary items 
print(student["name"])
print(student["age"])


#accessing with get method 
print(student.get('surname'))
print(student.get("name"))
```

```python
#single key is always used 
student_1={"name":"Dakshit",
           "age":23,
           "name":"chandel"}
print(student_1)
```

```python
##modifying dictonary elements
##dictonary are mutable , so you can add , update or delete element 
print(student)
```

```python
#updating a key value 
student["age"]=33
print(student)
```

```python
#adding new key value 
student["address"]="India"
print(student)
```

```python
#deleting a key value pair 
del student["age"]
print(student)

```
