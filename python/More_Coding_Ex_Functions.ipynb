```python
# Temperature
temp = int(input("Enter the temp you want to convert"))
unit = input("Enter the Unit")


def temp_convert(temp, unit):
    if unit == "C":
        return temp * 9 / 5 + 32
    elif unit == " F":
        return (temp - 32) * 5 / 9
    else:
        return None


print(temp_convert(temp, unit))
```
